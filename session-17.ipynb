```python
# Generate the feature vectors

import pandas as pd
from CBFV import composition

# Make a copy of the dataframe
df = df.copy()

# Remove rows where formula or target is missing
df = df.dropna(subset=['formula', 'target'])

# Convert formula column to string
df['formula'] = df['formula'].astype(str).str.strip()

# Remove empty formulas
df = df[df['formula'] != '']

# Convert target to numeric
df['target'] = pd.to_numeric(df['target'], errors='coerce')

# Remove rows with invalid target values
df = df.dropna(subset=['target'])

# Generate CBFV feature vectors
X, y, formulae, skipped = composition.generate_features(
    df,
    elem_prop='magpie',
    drop_duplicates=False,
    extend_features=True
)

print("Feature generation completed!")
print("X shape:", X.shape)
print("y shape:", y.shape)
print("Number of skipped formulas:", len(skipped))

print("\nFeature vectors:")
print(X.head())
```

```python
# 3. Inspect the Matrix
print(f"Feature matrix shape: {X.shape}")
```

```python
# 4. Check for all-zero features and missing values
zero_features = (X == 0).all().sum()
nan_count = X.isna().sum().sum()
```

```python
# 5. Export the validated feature matrix
X.to_csv('cbfv_features.csv', index=False)
print("Feature matrix successfully saved and ready for commit.")
```
